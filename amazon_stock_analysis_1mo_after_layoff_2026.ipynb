# Get Stock Information of Amazon, Google, Facebook, Microsoft, and NVIDIA

```python
import yfinance as yf
from datetime import date, timedelta

tickers = ['AMZN', 'GOOGL', 'META', 'MSFT', 'NVDA']

start_date = '2026-01-01'
end_date = '2026-02-28'

data = yf.download(tickers, start=start_date, end=end_date)

close_price = data['Close']

close_price.head()
```

# Visualize the Data

```python
# Normalize the stock price
# (calculating the rate of change with the start date as 1.0)

import pandas as pd
import matplotlib.pyplot as plt

normalized_data = close_price / close_price.iloc[0]


# Visualize the data

plt.figure(figsize=(12, 6))
for ticker in tickers:
    plt.plot(normalized_data[ticker], label=ticker)

# Adding a line to show the lay off date


plt.axvline(pd.Timestamp('2026-01-28'), color='red', linestyle='--', alpha=0.7, label='Amazon Layoff')

plt.title('Stock Price Change Comparison (Jan - Feb 2026)')
plt.xlabel('Date')
plt.ylabel('Normalized Price (Start = 1.0)')
plt.legend()
plt.grid(True)
plt.savefig('stock_comparison.png', bbox_inches='tight', dpi=300)
plt.show()


```
